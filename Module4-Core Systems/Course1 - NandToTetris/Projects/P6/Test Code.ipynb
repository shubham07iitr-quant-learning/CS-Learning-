```python
def instructionType(instruction):
    if instruction[0] == "(" and instruction[-1] == ")":
        return "A_TYPE"
    if instruction[0] == "@":
        return "B_TYPE"
    else:
        return "C_TYPE"


def cdecode(instruction):
    if instructionType(instruction) != "C_TYPE":
        raise ValueError("Pls input only C type instructions")
    else:
        leninstruction = len(instruction)
        equalindex = None 
        colonindex = None
        for i in range(leninstruction):
            if instruction[i] == ";":
                colonindex = i 
            else:
                if instruction[i] == "=":
                    equalindex = i
        if colonindex and equalindex:
            jumpstr = instruction[colonindex+1:]
            deststr = instruction[0: equalindex]
            compstr = instruction[equalindex+1:colonindex]
        if colonindex and equalindex == None:
            deststr = None
            compstr = instruction[0:colonindex]
            jumpstr = instruction[colonindex+1:]
        if equalindex and colonindex == None:
            jumpstr = None
            deststr = instruction[0: equalindex]
            compstr = instruction[equalindex+1:]
        if equalindex ==None and colonindex ==None:
            jumpstr = None 
            deststr = None 
            compstr = instruction

        return (deststr, compstr, jumpstr)
```

```python
a = "D=D+1;JMP"
x = "0;JMP"
y = "D=D+1"
z = "1"
```

```python
b = instructionType(a)
print(b)
```

```python
c = cdecode(a)
d = cdecode(x)
e = cdecode(y)
f = cdecode(z)
print(c)
print(d)
print(e)
print(f)
```

```python
def firstpass(listinstructiontuple, symboldict):
    for i in range(len(listinstructiontuple)):
        if listinstructiontuple[i][-1] == "L_TYPE":
            symboldict[listinstructiontuple[i][0]] = i
```
