```python
def solveForSingleCommentAndNewLine(finalList):
    finalListFinal = []
    for i in finalList:
        matches =re.search(r"(.*)[/][/](.*)", i)
        if matches:
            first, last = matches.groups()
            if first.strip() != "":
                finalListFinal.append(first.strip())
        elif i == "\n":
            continue
        elif i[-1:] == "\n":
            stripped = i[:-1].strip()
            if stripped != "":
                finalListFinal.append(stripped)
        else:
            finalListFinal.append(i.strip())
    return finalListFinal
        
        
```

# Finally we need to solve for the case when we have split the results by spaces but need to reconfigure if there is a string constant

```python
a = ['shubham', '"shubham gupta"', "survi"]
```

```python
a
```

```python
def reconfigureStringConstant(listFinal):
    finalListFinal = []
    waitForConfigure = False
    reconfiguredWord = ""
    for i in listFinal:
        if waitForConfigure:
            if i[-1] == '"':
                reconfiguredWord = reconfiguredWord + " " + i[0:-1]
                finalListFinal.append(reconfiguredWord)
                waitForConfigure = False
                reconfiguredWord = ""
            else:
                reconfiguredWord = reconfiguredWord + " " + i
                continue
        elif i[0] == '"':
            reconfiguredWord = i[1:]
            waitForConfigure = True
        else:
            finalListFinal.append(i)
    return finalListFinal            
```
